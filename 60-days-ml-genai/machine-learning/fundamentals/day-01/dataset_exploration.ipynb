Task 5: Training and testing split
Use Scikit-learn's train_test_split.

Requirements:
- Training dataset: 75%.
- Testing dataset: 25%.
- random_state=42.
- Use stratify to preserve class proportions approximately.

```python
import sklearn

print(sklearn.__version__)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
X = df.drop(columns=['customer_id', 'churned'])
y = df['churned']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
X_train.head()
```

```python
contract_type_encoded = le.fit_transform(X_train['contract_type'])
contract_type_encoded
```

```python
X_train['contract_type_encoded'] = contract_type_encoded
X_train.drop(columns=['contract_type'], inplace=True)
```

```python
X_test['contract_type_encoded'] = le.transform(X_test['contract_type'])
X_test.drop(columns=['contract_type'], inplace=True)
```
